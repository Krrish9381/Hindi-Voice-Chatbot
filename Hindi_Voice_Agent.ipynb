```python
import whisper
import os
import gradio as gr

from langchain_nvidia_ai_endpoints import ChatNVIDIA
```

```python
whisper_model = whisper.load_model("medium")
```

```python
os.environ["NVIDIA_API_KEY"] = "YOUR_NVIDIA_API_KEY"

llm = ChatNVIDIA(
    model="nvidia/nemotron-3-ultra-550b-a55b",
    api_key=os.environ["NVIDIA_API_KEY"]
)
```

```python
def speech_to_text(audio):

    result = whisper_model.transcribe(
        audio,
        language="hi"
    )

    return result["text"]
```

```python
def ask_llm(text):

    answer = llm.invoke(text)

    return answer.content
```

```python
def chatbot(audio):

    transcript = speech_to_text(audio)

    response = ask_llm(transcript)

    return transcript, response
```

```python
import gradio as gr

demo = gr.Interface(
    fn=chatbot,
    inputs=gr.Audio(
        sources=["microphone"],
        type="filepath",
        label="Speak in Hindi"
    ),
    outputs=[
        gr.Textbox(label="Transcript"),
        gr.Textbox(label="LLM Response")
    ],
    title="Hindi Voice Assistant"
)

demo.launch(debug=True)
```

```python
def chatbot(audio):
    transcript = speech_to_text(audio)
    response = ask_llm(transcript)

    # convert response -> speech

    return transcript, response, "output.mp3"
```

```python
import edge_tts
import asyncio
```

```python
async def text_to_speech(text):
    communicate = edge_tts.Communicate(
        text=text,
        voice="hi-IN-SwaraNeural"   # Hindi female voice
    )
    await communicate.save("output.mp3")
```

```python
def chatbot(audio):

    transcript = speech_to_text(audio)

    response = ask_llm(transcript)

    asyncio.run(text_to_speech(response))

    return transcript, response, "output.mp3"
```

```python
import gradio as gr

demo = gr.Interface(
    fn=chatbot,
    inputs=gr.Audio(
        sources=["microphone"],
        type="filepath",
        label="Speak in Hindi"
    ),
    outputs=[
        gr.Textbox(label="Transcript"),
        gr.Textbox(label="LLM Response"),
        gr.Audio(label="Hindi Speech")
    ],
    title="Hindi Voice Assistant"
)

demo.launch(debug=True)
```
